# Eliminace debilů: swap jmen a příjmení

```python
# Funkce pro kontrolu a prohození křestního jména a příjmení
def kontrola_a_prohozeni(row):
    prijmeni = row['Příjmení']
    krestni_jmeno = row['Křestní jméno']

    # Případ, kdy křestní jméno je ve sloupci "Příjmení" a naopak
    if prijmeni in seznam_jmen and krestni_jmeno not in seznam_jmen:
        return pd.Series([krestni_jmeno, prijmeni], index=['Příjmení', 'Křestní jméno'])
    elif krestni_jmeno in seznam_jmen and prijmeni not in seznam_jmen:
        return pd.Series([prijmeni, krestni_jmeno], index=['Příjmení', 'Křestní jméno'])
    else:
        return pd.Series([prijmeni, krestni_jmeno], index=['Příjmení', 'Křestní jméno'])
```

```python
# Aplikace funkce na celý DataFrame
df[['Příjmení', 'Křestní jméno']] = df.apply(kontrola_a_prohozeni, axis=1)
```

# Eliminace debilů: Přenesení striků

```python
# 2) Funkce pro doplnění strike1 a strike2 na základě existující hodnoty v rámci uživatele (e-mail)
def doplnit_striky(df):
    df['strike1'] = df.groupby('e-mail')['strike1'].transform(lambda x: x.max())
    df['strike2'] = df.groupby('e-mail')['strike2'].transform(lambda x: x.max())
    return df
```

```python
# Aplikujeme funkci pro doplnění strike1 a strike2
df = doplnit_striky(df)
```

```python
# 3) Převod sloupců strike1 a strike2 na bool
df['strike1'] = df['strike1'].apply(lambda x: True if str(x).lower() == 'true' else False)
df['strike2'] = df['strike2'].apply(lambda x: True if str(x).lower() == 'true' else False)
```

```python
# 4) Funkce pro filtrování a odstranění záznamů starších než zadané datum
def drop_stare_zaznamy(df):
    # Zeptáme se uživatele na datum ve formátu dd-mm-yyyy
    zadane_datum = input("Zadejte datum ve formátu DD-MM-YYYY: ")

    # Převod zadaného data na formát datetime
    try:
        zadane_datum = pd.to_datetime(zadane_datum, format="%d-%m-%Y")
    except ValueError:
        print("Zadané datum není platné. Prosím, použijte formát DD-MM-YYYY.")
        return df

    # Odstraníme všechny záznamy starší než zadané datum
    df_filtered = df[df['Datum'] >= zadane_datum]

    return df_filtered
```

```python
# Funkce pro anonymizaci e-mailu
def anonymizovat_email(email):
    try:
        local_part, domain = email.split('@')
    except ValueError:
        return email  # Pokud není validní e-mailová adresa, vracíme původní hodnotu
    if len(local_part) > 2:
        anonymized_local = local_part[0] + '*' * (len(local_part) - 2) + local_part[-1]
    else:
        anonymized_local = local_part  # Pokud je lokální díl krátký, neanonymizujeme
    return anonymized_local + '@' + domain

# Aplikace anonymizace e-mailů na celý DataFrame df_filtered
df_filtered['e-mail'] = df_filtered['e-mail'].apply(anonymizovat_email)

# Seznamy uživatelů s jejich příjmeními, křestními jmény a anonymizovanými e-maily
df_no_strikes['e-mail'] = df_no_strikes['e-mail'].apply(anonymizovat_email)
df_one_strike['e-mail'] = df_one_strike['e-mail'].apply(anonymizovat_email)
df_two_strikes['e-mail'] = df_two_strikes['e-mail'].apply(anonymizovat_email)

# Vytvoření seznamů s anonymizovanými e-maily
list_no_strikes = df_no_strikes[['Příjmení', 'Křestní jméno', 'e-mail']].values.tolist()
list_one_strike = df_one_strike[['Příjmení', 'Křestní jméno', 'e-mail']].values.tolist()
list_two_strikes = df_two_strikes[['Příjmení', 'Křestní jméno', 'e-mail']].values.tolist()

# Funkce pro vytvoření dvojic nebo trojic s kruhovým propojením
def vytvorit_dvojice(data):
    result = []
    for i in range(0, len(data), 2):
        if i + 2 < len(data):
            result.append(data[i:i+2])
        elif i + 2 == len(data):
            result.append(data[i:i+2])
        elif i + 1 == len(data):
            last_two = result[-1]
            result[-1] = [last_two[0], last_two[1], data[i]]  # vytvoření trojice
    return result

# Funkce pro vytvoření tabulky s kruhovými trojicemi
def vytvorit_dvojradkovou_tabulku(dvojice):
    rows = []
    for group in dvojice:
        if len(group) == 2:
            rows.append([group[0][0], group[0][1], group[0][2], group[1][0], group[1][1], group[1][2]])
            rows.append([group[1][0], group[1][1], group[1][2], group[0][0], group[0][1], group[0][2]])
        elif len(group) == 3:
            rows.append([group[0][0], group[0][1], group[0][2], group[1][0], group[1][1], group[1][2]])
            rows.append([group[1][0], group[1][1], group[1][2], group[2][0], group[2][1], group[2][2]])
            rows.append([group[2][0], group[2][1], group[2][2], group[0][0], group[0][1], group[0][2]])
    return rows

# Funkce pro kontrolu prázdných seznamů a vytvoření tabulek
def vytvorit_a_zobrazit_tabulku(dvojice, nadpis):
    if dvojice:
        tabulka_dup = vytvorit_dvojradkovou_tabulku(dvojice)
        df_dup = pd.DataFrame(tabulka_dup, columns=['Příjmení 1', 'Křestní jméno 1', 'E-mail 1', 'Příjmení 2', 'Křestní jméno 2', 'E-mail 2'])
        print(nadpis)
        print(df_dup.sort_values('Příjmení 1').to_string(index=False))  # Zajišťuje správné tisknutí místo display()
    else:
        print(f"Žádní uživatelé v kategorii: {nadpis}")

# Vytvoření dvojřádkových tabulek pro každou kategorii
vytvorit_a_zobrazit_tabulku(vytvorit_dvojice(list_no_strikes), "Tabulka uživatelů bez strike")
vytvorit_a_zobrazit_tabulku(vytvorit_dvojice(list_one_strike), "\nTabulka uživatelů s jedním strike")
vytvorit_a_zobrazit_tabulku(vytvorit_dvojice(list_two_strikes), "\nTabulka uživatelů se dvěma striky")
```
